```python
import pandas as pd
from sqlalchemy import create_engine, text
from sqlalchemy.engine import URL

print("SQL Analysis notebook started successfully!")
```

```python
username = "root"
password = "Srivani@123"

connection_url = URL.create(
    drivername="mysql+pymysql",
    username=username,
    password=password,
    host="localhost",
    port=3306,
    database="methanol_project"
)

engine = create_engine(connection_url)

with engine.connect() as connection:
    print("Connected database:", connection.execute(
        text("SELECT DATABASE()")
    ).scalar())
```

```python
query = """
SELECT COUNT(*) AS Total_Records
FROM equilibrium_results;
"""

record_count = pd.read_sql(text(query), engine)
display(record_count)
```

```python
query = """
SELECT
    Temperature_K,
    Pressure_bar,
    CO2_Conversion_pct,
    Methanol_Yield_pct
FROM equilibrium_results
ORDER BY Methanol_Yield_pct DESC
LIMIT 10;
"""

top_conditions = pd.read_sql(text(query), engine)
display(top_conditions.round(3))
```

```python
query = """
SELECT
    Pressure_bar,
    ROUND(AVG(Methanol_Yield_pct), 2) AS Avg_Methanol_Yield
FROM equilibrium_results
GROUP BY Pressure_bar
ORDER BY Pressure_bar;
"""

pressure_analysis = pd.read_sql(text(query), engine)
display(pressure_analysis)
```

```python
query = """
SELECT
    Temperature_K,
    ROUND(AVG(Methanol_Yield_pct), 2) AS Avg_Methanol_Yield
FROM equilibrium_results
GROUP BY Temperature_K
ORDER BY Temperature_K;
"""

temperature_analysis = pd.read_sql(text(query), engine)
display(temperature_analysis)
```

```python
query = """
SELECT
    Temperature_K,
    Pressure_bar,
    ROUND(Methanol_Yield_pct, 2) AS Methanol_Yield_pct
FROM equilibrium_results
ORDER BY Temperature_K, Pressure_bar;
"""

yield_comparison = pd.read_sql(text(query), engine)
display(yield_comparison)
```

```python
query = """
SELECT
    Temperature_K,
    Pressure_bar,
    CO2_Conversion_pct,
    Methanol_Yield_pct,
    Methanol_mol
FROM equilibrium_results
ORDER BY Methanol_Yield_pct DESC
LIMIT 1;
"""

best_condition = pd.read_sql(text(query), engine)
display(best_condition.round(3))
```

```python
query = """
SELECT
    Solver_success,
    COUNT(*) AS Number_of_Cases
FROM equilibrium_results
GROUP BY Solver_success;
"""

solver_check = pd.read_sql(text(query), engine)
display(solver_check)
```

```python
query = """
SELECT
    Solver_success,
    COUNT(*) AS Number_of_Cases
FROM equilibrium_results
GROUP BY Solver_success
ORDER BY Solver_success;
"""

solver_check = pd.read_sql(text(query), engine)
print(solver_check.to_string(index=False))
```

```python
query = """
SELECT
    COUNT(*) AS Total_Cases,
    ROUND(AVG(CO2_Conversion_pct), 2) AS Avg_CO2_Conversion_pct,
    ROUND(MAX(CO2_Conversion_pct), 2) AS Max_CO2_Conversion_pct,
    ROUND(AVG(Methanol_Yield_pct), 2) AS Avg_Methanol_Yield_pct,
    ROUND(MAX(Methanol_Yield_pct), 2) AS Max_Methanol_Yield_pct
FROM equilibrium_results
WHERE Solver_success = 1;
"""

performance_summary = pd.read_sql(text(query), engine)
display(performance_summary)
```

```python
query = """
SELECT
    Pressure_bar,
    ROUND(AVG(Methanol_Yield_pct), 2) AS Avg_Yield_pct,
    ROUND(MAX(Methanol_Yield_pct), 2) AS Max_Yield_pct
FROM equilibrium_results
WHERE Solver_success = 1
GROUP BY Pressure_bar
ORDER BY Avg_Yield_pct DESC;
"""

pressure_ranking = pd.read_sql(text(query), engine)
display(pressure_ranking)
```

```python
query = """
SELECT
    Temperature_K,
    ROUND(AVG(Methanol_Yield_pct), 2) AS Avg_Yield_pct,
    ROUND(MAX(Methanol_Yield_pct), 2) AS Max_Yield_pct
FROM equilibrium_results
WHERE Solver_success = 1
GROUP BY Temperature_K
ORDER BY Avg_Yield_pct DESC;
"""

temperature_ranking = pd.read_sql(text(query), engine)
display(temperature_ranking)
```

```python
query = """
SELECT
    Temperature_K,
    Pressure_bar,
    ROUND(Methanol_Yield_pct, 2) AS Methanol_Yield_pct
FROM equilibrium_results
WHERE Solver_success = 1
ORDER BY Temperature_K, Pressure_bar;
"""

yield_data = pd.read_sql(text(query), engine)

yield_matrix = yield_data.pivot(
    index="Temperature_K",
    columns="Pressure_bar",
    values="Methanol_Yield_pct"
)

display(yield_matrix)
```

```python
yield_data.to_csv("methanol_yield_analysis.csv", index=False)
yield_matrix.to_csv("methanol_yield_matrix.csv")

print("Analysis files exported successfully!")
```

```python
query = """
SELECT
    COUNT(*) AS Total_Cases,
    ROUND(AVG(CO2_Conversion_pct), 2) AS Average_Conversion_pct,
    ROUND(MAX(CO2_Conversion_pct), 2) AS Maximum_Conversion_pct,
    ROUND(AVG(Methanol_Yield_pct), 2) AS Average_Yield_pct,
    ROUND(MAX(Methanol_Yield_pct), 2) AS Maximum_Yield_pct
FROM equilibrium_results
WHERE Solver_success = 1;
"""

final_summary = pd.read_sql(text(query), engine)
display(final_summary)
```
